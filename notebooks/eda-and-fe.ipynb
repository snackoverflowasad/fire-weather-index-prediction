```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
```

```python
df = pd.read_csv('data/Algerian_forest_fires_dataset_UPDATE.csv', header=1)
```

```python
df.head()
```

```python
df.info()
```

```python
df[df.isnull().any(axis=1)]
```

```python
df.loc[:122, "Region"] = 0
df.loc[122:, "Region"] = 1
```

```python
df.info()
```

```python
df['Region'] = df['Region'].astype('int64')
```

```python
df.info()
df.isnull().sum()
```

```python
df = df.dropna().reset_index(drop=True)
```

```python
df.isnull().sum()
```

```python
df = df.drop(122).reset_index(drop=True)
```

```python
df.iloc[[122]]
```

```python
df.columns
```

```python
df.columns = df.columns.str.strip()
```

```python
df.info()
```

```python
df[['day', 'month', 'year', 'Temperature', 'RH', 'Ws']] = df[['day', 'month', 'year', 'Temperature', 'RH', 'Ws']].astype('int64')
```

```python
df.info()
```

```python
df.head()
```

```python
objects = [features for features in df.columns if df[features].dtype=='O']
```

```python
for obj in objects:
    if obj != 'Classes':
        df[obj] = df[obj].astype('float64')
```

```python
df.info()
```

```python
df.describe()
```

```python
df.to_csv('data/Algerian_forest_fires_dataset_cleaned.csv', index=False)
```
